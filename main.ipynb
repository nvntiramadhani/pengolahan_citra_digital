```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
nama_file = "red flower.jpg"

gambar = cv2.imread(nama_file)

if gambar is None:
    raise ValueError("Gambar gagal dibaca!")
```

```python
# Mengubah format warna BGR menjadi RGB
gambar_rgb = cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB)

# Menampilkan gambar asli
plt.figure(figsize=(6, 8))
plt.imshow(gambar_rgb)
plt.title("Citra Asli - Bunga Mawar")
plt.axis("off")
plt.show()
```

```python
# Mengubah citra berwarna menjadi grayscale
gambar_gray = cv2.cvtColor(gambar, cv2.COLOR_BGR2GRAY)

print("Proses grayscale berhasil dilakukan!")
```

```python
# Menampilkan perbandingan citra asli dan grayscale
plt.figure(figsize=(12, 7))

# Menampilkan citra asli
plt.subplot(1, 2, 1)
plt.imshow(gambar_rgb)
plt.title("Citra Asli - Bunga Mawar")
plt.axis("off")

# Menampilkan citra grayscale
plt.subplot(1, 2, 2)
plt.imshow(gambar_gray, cmap="gray", vmin=0, vmax=255)
plt.title("Hasil Grayscale")
plt.axis("off")

# Mengatur tata letak gambar
plt.tight_layout()
plt.show()
```

```python
# Menentukan nama file hasil
nama_output = "hasil_grayscale_mawar.png"

# Menyimpan citra grayscale
berhasil = cv2.imwrite(nama_output, gambar_gray)

# Memeriksa keberhasilan penyimpanan
if berhasil:
    print("Citra grayscale berhasil disimpan!")
    print("Nama file:", nama_output)
else:
    print("Gagal menyimpan citra grayscale.")
```

```python
# Membaca kembali file hasil grayscale
hasil_simpan = cv2.imread(
    nama_output,
    cv2.IMREAD_GRAYSCALE
)

# Memeriksa apakah file berhasil dibaca
if hasil_simpan is not None:
    plt.figure(figsize=(6, 8))
    plt.imshow(hasil_simpan, cmap="gray", vmin=0, vmax=255)
    plt.title("Hasil Grayscale yang Disimpan")
    plt.axis("off")
    plt.show()

    print("File hasil berhasil dibuka kembali!")
    print("Nama file:", nama_output)
    print("Lokasi file:", os.path.abspath(nama_output))
else:
    print("File hasil tidak ditemukan atau gagal dibaca.")
```
